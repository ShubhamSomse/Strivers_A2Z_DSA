What this pattern asks

For every element in an array, find:

Nearest Greater / Smaller on the LEFT

Nearest Greater / Smaller on the RIGHT

“Nearest” means closest, not any greater/smaller.

The 4 possible variants (you don’t memorize separately)
You want	Stack type
Next Greater (Right)	Monotonic Decreasing
Previous Greater (Left)	Monotonic Decreasing
Next Smaller (Right)	Monotonic Increasing
Previous Smaller (Left)	Monotonic Increasing
